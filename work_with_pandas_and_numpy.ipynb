```python
import numpy as np
scores=np.array([80,95,87,66,90])
print(scores)

zeros=np.zeros(5)
print(zeros)

ones=np.ones(20)
print(ones)

empty=np.empty(5)
print(empty)

range=np.arange(1,10,2)
print(range)

cur_score=scores+5
print(cur_score)
print(np.mean(scores))
print(np.median(scores))
print(np.max(scores))
```

**panda**

```python
import pandas as pd

# Creating a DataFrame using a dictionary
data = {
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Age": [25, 30, 35, 28],
    "Salary": [50000, 60000, 75000, 52000],
    "Department": ["HR", "IT", "IT", "Sales"]
}

df = pd.DataFrame(data)
print(df)
```

```python
print(df.head(2))      # Shows the first 2 rows
print(df.tail(2))      # Shows the last 2 rows
print(df.shape)        # Shows (Number of Rows, Number of Columns) -> (4, 4)
print(df.info())       # Shows data types and if there are missing values
print(df.describe())   # Gives quick math stats (mean, min, max) for number columns
```

```python
print(df["Name"])
print(df[["Name","Salary"]])
print(df.iloc[0])
print(df[df["Department"]=="IT"])
print(df[df["Age"]>28])
avg_salary=df["Salary"].mean()
print(avg_salary)
```
