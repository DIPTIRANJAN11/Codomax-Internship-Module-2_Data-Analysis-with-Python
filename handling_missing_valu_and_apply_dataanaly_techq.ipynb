```python
import pandas as pd
import numpy as np

# Create a dataset with missing values
data = {
    "Product": ["Laptop", "Mouse", "Keyboard", "Monitor", "Webcam"],
    "Price": [1000, 25, np.nan, 200, 50],
    "Units_Sold": [10, np.nan, 50, 15, np.nan],
    "Rating": [4.5, 4.0, 3.8, np.nan, 4.2]
}
df = pd.DataFrame(data)

print("--- Raw Data with Missing Values ---")
print(df)

# 1. FILL missing values with the AVERAGE
df["Price"] = df["Price"].fillna(df["Price"].mean())

# 2. FILL missing values with a SPECIFIC number (e.g., 0 sales means it wasn't sold)
df["Units_Sold"] = df["Units_Sold"].fillna(0)

# 3. DROP rows where the Rating is missing (we don't want to guess ratings)
df = df.dropna(subset=["Rating"])

print("\n--- Cleaned Data ---")
print(df)
```

```python
print("\n--- Basic Data Analysis ---")

# 1. DESCRIPTIVE STATISTICS
print("Summary Statistics of the dataset:")
print(df.describe()) # Instantly shows mean, min, max, and percentiles for all number columns

# 2. SIMPLE MATH (Total Revenue)
# Create a new column by multiplying Price by Units_Sold
df["Total_Revenue"] = df["Price"] * df["Units_Sold"]
print("\nData with Total Revenue added:")
print(df)

# 3. AGGREGATION
total_store_revenue = df["Total_Revenue"].sum()
print(f"\nTotal Store Revenue: ${total_store_revenue}")

best_selling_product = df.loc[df["Total_Revenue"].idxmax(), "Product"]
print(f"Best Selling Product (by revenue): {best_selling_product}")

# 4. CORRELATION
# Checks if higher prices relate to higher ratings (Values close to 1 mean strong positive relation)
correlation = df["Price"].corr(df["Rating"])
print(f"\nCorrelation between Price and Rating: {correlation:.2f}")
```

```python
import pandas as pd
import numpy as np

# ==========================================
# 1. CREATE MESSY DATA
# ==========================================
data = {
    "Product": ["Laptop", "Mouse", "Keyboard", "Monitor", "Webcam", "Tablet", "Printer"],
    "Price": [1000, 25, np.nan, 200, 50, 300, np.nan],
    "Units_Sold": [10, np.nan, 50, 15, np.nan, 20, 5],
    "Customer_Rating": [4.5, 4.0, 3.8, np.nan, 4.2, 4.8, 3.0]
}
df = pd.DataFrame(data)

print("=== RAW DATA ===")
print(df)
print("\n")

# ==========================================
# 2. HANDLING MISSING VALUES
# ==========================================
print("=== HANDLING MISSING VALUES ===")

# Fill missing Price with the average price of other items
avg_price = df["Price"].mean()
df["Price"] = df["Price"].fillna(avg_price)
print(f"Filled missing Prices with the average (${avg_price:.2f})")

# Fill missing Units Sold with 0 (assume no sales)
df["Units_Sold"] = df["Units_Sold"].fillna(0)
print("Filled missing Units Sold with 0")

# Drop the row where Customer Rating is missing
# (We don't want to use a fake rating for our analysis)
df = df.dropna(subset=["Customer_Rating"])
print("Dropped rows with missing Customer Ratings")
print("\n")


# ==========================================
# 3. BASIC DATA ANALYSIS
# ==========================================
print("=== DATA ANALYSIS ===")

# A) Calculated Columns
df["Total_Revenue"] = df["Price"] * df["Units_Sold"]

# B) Descriptive Statistics
print("1. Summary Statistics for the Store:")
print(df[["Price", "Units_Sold", "Total_Revenue"]].describe().round(2))

# C) Finding Extremes (Max and Min)
highest_revenue_product = df.loc[df["Total_Revenue"].idxmax(), "Product"]
print(f"\n2. Product with the Highest Revenue: {highest_revenue_product}")

# D) Grouping (Aggregation) & Sorting
print("\n3. Average Rating by Product (Sorted Highest to Lowest):")
print(df.groupby("Product")["Customer_Rating"].mean().sort_values(ascending=False))

# E) Data Filtering
print("\n4. Premium Products (Price > $100):")
premium_products = df[df["Price"] > 100]
print(premium_products[["Product", "Price", "Total_Revenue"]])

print("\n=== FINAL CLEANED DATA ===")
print(df)
```
