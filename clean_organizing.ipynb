```python
import pandas as pd

# Let's create a "messy" mock dataset to practice on
data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Alice"], # Duplicate Alice
    "Age": [25, None, 30, 22, 25],                         # Missing Age for Bob
    "City": ["New York", "London", "new york", "Paris", "New York"], # Messy text
    "Salary": [50000, 60000, 75000, 48000, 50000]
}
df = pd.DataFrame(data)

# --- EXPLORING THE DATA ---
print("1. Shape of the dataset (Rows, Columns):", df.shape)
print("\n2. Column Names:", df.columns)
print("\n3. Data Types and Missing Values:")
print(df.info()) # Shows if columns have Null/None values
```

```python
# --- CLEANING THE DATA ---

# 1. Handling Missing Values
# Option A: Drop rows with missing data (df.dropna())
# Option B: Fill missing data with a value (like the average)
df["Age"] = df["Age"].fillna(df["Age"].mean()) # Fills Bob's missing age with the average (25.6)

# 2. Removing Duplicates
df = df.drop_duplicates()

# 3. Fixing Text (Standardizing formats)
# Notice "new york" and "New York". Let's make the first letter capital for all words.
df["City"] = df["City"].str.title()

print("\n--- Cleaned Data ---")
print(df)
```

```python
# --- ORGANIZING THE DATA ---

# 1. Renaming columns (Making them lowercase or more descriptive)
df = df.rename(columns={"Name": "Employee_Name", "Salary": "Annual_Salary"})

# 2. Sorting data (e.g., Highest salary first)
df = df.sort_values(by="Annual_Salary", ascending=False)

# 3. Resetting the index numbers so they are back in order (0, 1, 2...)
df = df.reset_index(drop=True)

print("\n--- Final Organized Data ---")
print(df)
```

```python
import pandas as pd
import numpy as np

# 1. CREATE MESSY DATA
data = {
    "CustomerID": [101, 102, 103, 104, 101], # Duplicate ID
    "Name": ["Alice", "Bob", "Charlie", "David", "Alice"],
    "Age": [25, np.nan, 30, 22, 25],         # np.nan means "Missing Value"
    "City": ["  New York ", "London", "new york", " Paris", "New York"],
    "Purchase": [500.00, 150.00, None, 250.00, 500.00]
}
df = pd.DataFrame(data)

print("=== RAW (MESSY) DATA ===")
print(df)
print("\n")

# 2. EXPLORE DATA
print("=== EXPLORING DATA ===")
print(f"Total Rows: {df.shape[0]}, Total Columns: {df.shape[1]}")
print("Missing values per column:")
print(df.isnull().sum()) # Counts missing values in each column
print("\n")


# 3. CLEAN DATA
print("=== CLEANING DATA ===")

# a) Remove duplicates based on CustomerID
df = df.drop_duplicates(subset=["CustomerID"], keep="first")
print("Duplicates removed.")

# b) Fix messy text (Remove spaces and standardize capitalization)
df["City"] = df["City"].str.strip().str.title()
print("City names formatted.")

# c) Handle missing numbers (Fill with the average of the column)
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Purchase"] = df["Purchase"].fillna(0) # Assume missing purchase = $0
print("Missing values handled.")
print("\n")


# 4. ORGANIZE DATA
print("=== ORGANIZING DATA ===")

# a) Rename columns for clarity
df = df.rename(columns={"Purchase": "Total_Spent", "CustomerID": "ID"})

# b) Sort by highest spender
df = df.sort_values(by="Total_Spent", ascending=False)

# c) Reset index so it looks clean
df = df.reset_index(drop=True)

print("=== FINAL CLEAN & ORGANIZED DATA ===")
print(df)
```
